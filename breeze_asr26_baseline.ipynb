```python
# 下載並載入模型
import torch
from transformers import pipeline

if not torch.cuda.is_available():
    raise RuntimeError("沒有偵測到 GPU，請先將 Colab 執行階段切換成 GPU。")

model_id = "MediaTek-Research/Breeze-ASR-26"

asr = pipeline(
    "automatic-speech-recognition",
    model=model_id,
    device=0,
    dtype=torch.float16,
    chunk_length_s=30, # 啟用自動分段辨識，可接收長音檔
)

print("模型載入完成！（已啟用長音檔分段辨識支援）")
```

```python
# 選一個音檔並執行辨識
from google.colab import files
from pathlib import Path
from time import perf_counter

uploaded = files.upload()

if not uploaded:
    raise ValueError("請先選擇一個錄音檔。")

audio_path = next(iter(uploaded))
Path(audio_path).write_bytes(uploaded[audio_path])

start = perf_counter()
# 辨識時加入 return_timestamps=True 以支援長音檔辨識
result = asr(audio_path, return_timestamps=True)
elapsed = perf_counter() - start

print("音檔：", audio_path)
print("辨識結果：", result["text"])
print(f"辨識耗時：{elapsed:.2f} 秒（不含上傳與模型下載）")
```

```python
# 測試網站上給的音檔
from huggingface_hub import hf_hub_download
from pathlib import Path
import tarfile

archive = hf_hub_download(
    repo_id="adi-gov-tw/Taiwan-Tongues-ASR-CE-dataset-hokkien",
    repo_type="dataset",
    filename="train/train-000000.tar",
)

samples = {
    "train-000000001": ("01.mp3", "食薰"),
    "train-000000002": ("02.mp3", "穩當"),
    "train-000000005": ("03.mp3", "只有孤單陪伴我"),
}

with tarfile.open(archive) as tar:
    for item in tar.getmembers():
        p = Path(item.name)
        if p.suffix.lower() == ".mp3" and p.stem in samples:
            filename, _ = samples[p.stem]
            Path(filename).write_bytes(tar.extractfile(item).read())

for filename, reference in samples.values():
    if not Path(filename).exists():
        print(f"找不到音檔：{filename}")
        continue
    print(f"\n音檔：{filename}")
    print("參考文字：", reference)
    print("模型輸出：", asr(filename)["text"])
```

```python
# 終極相容修正版：最純淨 16kHz WAV 格式轉換，徹底避免 Soundfile 損毀報錯
import subprocess
import os
from google.colab.output import eval_js
from base64 import b64decode
from pathlib import Path
from IPython.display import Audio, display

if "asr" not in globals():
    raise RuntimeError("請先執行前面載入 Breeze-ASR-26 的那一格。")

record_js = r"""
new Promise((resolve) => {
  const box = document.createElement('div');
  box.style.border = '1px solid #ccc';
  box.style.padding = '15px';
  box.style.borderRadius = '10px';
  box.style.width = '350px';
  box.style.background = '#f9f9f9';

  const start = document.createElement('button');
  const stop = document.createElement('button');
  const status = document.createElement('p');

  // 音量視覺化畫布
  const canvas = document.createElement('canvas');
  canvas.width = 300;
  canvas.height = 20;
  canvas.style.display = 'block';
  canvas.style.background = '#ddd';
  canvas.style.marginTop = '10px';
  canvas.style.borderRadius = '3px';

  start.textContent = '🎙 開始錄音';
  stop.textContent = '⏹ 停止並辨識';
  stop.disabled = true;
  status.textContent = '請點選「開始錄音」並允許麥克風權限。';

  box.append(start, stop, status, canvas);
  document.body.appendChild(box);

  let audioCtx;
  let analyser;
  let source;
  let animationId;
  let stream;
  let recorder;
  const chunks = [];

  function drawVolume() {
    const bufferLength = analyser.frequencyBinCount;
    const dataArray = new Uint8Array(bufferLength);
    analyser.getByteFrequencyData(dataArray);

    let sum = 0;
    for(let i = 0; i < bufferLength; i++) {
      sum += dataArray[i];
    }
    const average = sum / bufferLength;

    const ctx = canvas.getContext('2d');
    ctx.clearRect(0, 0, canvas.width, canvas.height);
    ctx.fillStyle = average > 5 ? '#2ecc71' : '#e74c3c';
    ctx.fillRect(0, 0, (average / 128) * canvas.width, canvas.height);

    animationId = requestAnimationFrame(drawVolume);
  }

  start.onclick = async () => {
    start.disabled = true;
    chunks.length = 0;
    try {
      stream = await navigator.mediaDevices.getUserMedia({audio: true});

      audioCtx = new (window.AudioContext || window.webkitAudioContext)();
      analyser = audioCtx.createAnalyser();
      source = audioCtx.createMediaStreamSource(stream);
      source.connect(analyser);
      analyser.fftSize = 256;
      drawVolume();

      recorder = new MediaRecorder(stream);
      recorder.ondataavailable = (event) => {
        if (event.data.size > 0) chunks.push(event.data);
      };

      recorder.onstop = () => {
        stream.getTracks().forEach(track => track.stop());
        cancelAnimationFrame(animationId);
        if (audioCtx) audioCtx.close();

        stop.disabled = true;
        status.textContent = '錄音結束！正在進行轉碼與辨識中...';

        const blob = new Blob(chunks, {type: recorder.mimeType});
        const reader = new FileReader();
        reader.onload = () => resolve({audio: reader.result});
        reader.onerror = () => resolve({error: '無法讀取錄音'});
        reader.readAsDataURL(blob);
      };

      recorder.start();
      stop.disabled = false;
      status.textContent = '🎙 錄音中，請對麥克風大聲講話...';
    } catch (error) {
      if (stream) stream.getTracks().forEach(track => track.stop());
      status.textContent = '❌ 錄音失敗：' + error.message;
      resolve({error: error.message});
    }
  };

  stop.onclick = () => {
    if (recorder && recorder.state === 'recording') {
      recorder.stop();
    }
  };
})
"""

recorded = eval_js(record_js)

if "error" in recorded:
    raise RuntimeError("錄音失敗：" + recorded["error"])

header, encoded = recorded["audio"].split(",", 1)
temp_webm = Path("temp_recorded.webm")
temp_webm.write_bytes(b64decode(encoded))

output_wav = "taigi_test.wav"

# 採用最安全穩定的常規轉換 (不套用任何容易出錯的過濾器)，保證檔案格式完全合規
cmd = f"ffmpeg -y -i {temp_webm} -ar 16000 -ac 1 {output_wav}"
subprocess.run(cmd, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

if temp_webm.exists():
    temp_webm.unlink()

# 安全性檢查
if not os.path.exists(output_wav) or os.path.getsize(output_wav) < 1000:
    raise RuntimeError("錄音檔案可能損毀或無訊號，請重試並大聲說話。")

# 執行辨識（套用防重複參數）
esult = asr(
    output_wav,
    generate_kwargs={
        "no_repeat_ngram_size": 4
    }
)

print("\n辨識結果：", result["text"] if result["text"].strip() else "（偵測為靜音或無效聲音）")

# 播放轉碼後的標準 WAV，此時一定可以播放
display(Audio(filename=output_wav))
```
